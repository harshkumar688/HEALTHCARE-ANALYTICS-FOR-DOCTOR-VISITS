```python
df = df.drop_duplicates()
```

```python
print(df['visits'].value_counts().sort_index())
```

```python
plt.figure(figsize=(10,5))

sns.countplot(data=df, x='visits')

plt.title("Distribution of Doctor Visits")
plt.xlabel("Number of Doctor Visits")
plt.ylabel("Number of Patients")

plt.show()
```

```python
df['visits'].describe()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df['age'], bins=15, kde=True)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df['income'], bins=15, kde=True)

plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(data=df, x='illness')

plt.title("Distribution of Number of Illnesses")
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df['health'], bins=15, kde=True)

plt.title("Health Condition Distribution")
plt.show()
```

```python
df.groupby('income')['visits'].agg(
    ['count', 'mean', 'median', 'sum']
).sort_index()
```

```python
plt.figure(figsize=(10,5))

sns.barplot(data=df, x='income', y='visits', estimator='mean')

plt.title("Average Doctor Visits by Income")
plt.xlabel("Income")
plt.ylabel("Average Visits")

plt.xticks(rotation=45)

plt.show()
```

```python
numerical_columns = [
    'visits',
    'age',
    'income',
    'illness',
    'reduced',
    'health'
]

for column in numerical_columns:

    plt.figure(figsize=(3,1))

    sns.boxplot(x=df[column])

    plt.title(f"Boxplot of {column}")

    plt.show()
```
